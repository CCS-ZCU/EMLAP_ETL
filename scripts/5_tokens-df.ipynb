```python
import pandas as pd
import os
import json
import spacy

from processing_textblocks_helpers import nlp_latin
```

```python
source_path = "../data/sents_data_jsons_dicts/"
filenames = os.listdir(source_path)
```

```python
len(filenames)
```

```python
emlap_tokens = []
for filename in filenames:
    f_sents_data = json.load(open(source_path + filename, "rb"))
    for sent_data in f_sents_data:
        sent_len = len(sent_data["sent_text"])
        for token_data in sent_data["tokens_data"]:
            token_data["id"] = sent_data["work_id"]
            token_data["sent_idx"] = sent_data["sent_id"]
            token_data["sent_len"] = sent_len
            emlap_tokens.append(token_data)
```

```python
len(emlap_tokens)
```

```python
emlap_tokens_df = pd.DataFrame(emlap_tokens)
```

```python
for attr in ["page", "textblock", "tag", "blocktype"]:
    emlap_tokens_df[attr] = emlap_tokens_df.apply(lambda row: row["ref"][attr], axis=1)
```

```python
path_sanitized_textblocks = "../data/emlap_sanitized_textblocks/"
path_raw_textblocks = "../data/emlap_annotated_textblocks/"

def show_raw_and_sanitized_textblock(id, page, textblock):
    filename = [f for f in os.listdir(path_sanitized_textblocks) if f[:6] == str(id)][0]
    with open(os.path.join(path_raw_textblocks, filename), 'r', encoding='utf-8') as f:
        textblocks_raw = json.load(f)
    with open(os.path.join(path_sanitized_textblocks, filename), 'r', encoding='utf-8') as f:
        textblocks_sanitized = json.load(f)
    print("filename: ", filename)
    print("page (0-based indexing): ", page)
    print("raw annotated textblock data: ", textblocks_raw[page][textblock])
    print("sanitized annotated textblock data: ", textblocks_sanitized[page][textblock])
```
