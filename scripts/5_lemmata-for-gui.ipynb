```python
emlaps_sents_data = []
for filename in filenames:
    f_sents_data = json.load(open(source_path + filename, "rb"))
    emlaps_sents_data.extend(f_sents_data)
```

```python
emlap_tokens = []
for sent_data in emlaps_sents_data:
    sent_len = len(sent_data[2])
    for token_data in sent_data[3]:
        if isinstance(token_data, list) and len(token_data) >= 6:
            emlap_tokens.append({
                "id": sent_data[0],
                "token": token_data[0],
                "lemma": token_data[1],
                "pos": token_data[2],
                "char_position": token_data[3],
                "sent_len": sent_len,  # length of the sentence
                "pdf_page": token_data[4],  # 0-based indexing!
                "pdf_page_textblock": token_data[5]
            })
        else:
            # Optionally log or print the offending entry for inspection
            print("Skipped malformed token_data:", token_data)
```

```python
emlap_tokens_df = pd.DataFrame(emlap_tokens)
```

```python
import numpy as np
import pandas as pd

def _flat_int_list(x):
    # turn scalars/arrays/Series/[arrays] into a flat list of ints
    if isinstance(x, (list, tuple)):
        seq = x
    elif isinstance(x, (np.ndarray, pd.Series)):
        seq = x.tolist()
    else:
        seq = [x]
    out = []
    for v in seq:
        if isinstance(v, (list, tuple, np.ndarray, pd.Series)):
            out.extend([int(i) for i in np.array(v).flatten().tolist()])
        else:
            out.append(int(v))
    return out

def get_positions(row):
    data = positions_data_dict[row["id"]]          # e.g. positions_data_dict[100044]
    pages  = _flat_int_list(row["pdf_page"])
    blocks = _flat_int_list(row["pdf_page_textblock"])

    # one page, many blocks → repeat page
    if len(pages) == 1 and len(blocks) > 1:
        pages = pages * len(blocks)

    out = []
    for p, b in zip(pages, blocks):
        page_blocks = data[p]                      # list of blocks on this page
        # simple 1-based fallback if needed
        if b >= len(page_blocks) and b > 0:
            b = b - 1
        # final guard: skip if still OOB
        if 0 <= b < len(page_blocks):
            out_el = page_blocks[b]
            out_el["page"] = p
            out.append(out_el)
        # else: silently skip (or append None if you prefer)
    return out

# apply
emlap_tokens_df["positions"] = emlap_tokens_df.apply(get_positions, axis=1)
```
